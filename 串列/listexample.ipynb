```python
scores = [78,29,56,80,99]
print(len(scores))
print(scores[4])
print(scores[5-1])
print(scores[len(scores)-1])
print(scores[-1])
```

使用`append()`新增元素串列的尾端

```python
print(scores)
scores.append(78)
scores.append(93)
print(scores)
```

```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
for i,e in enumerate(scores):
    print(i,e) 
```
