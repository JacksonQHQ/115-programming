字典,在Python 中用於查表。

• 字典的前後是用大括號 {} 包住

• 裏面每個元素用逗點（,）隔開

• 每個元素都是 key:value 的形式

```python
students = {'Peter': 175, 'Mary': 168, 'Keven': 165} 
# 查學生表格,使用關鍵字(key) Keven 來查
print(students['Keven'])
```

```python
students = {
'Peter': [175, 70, '高雄'],
'Mary': [168, 50, '嘉義'],
'Keven': [165, 55, '高雄']
}

print(students['Mary'])
print(students['Mary'][0]) #取得身高
print(students['Mary'][2]) #取得戶籍
```
