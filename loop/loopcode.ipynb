```python
for i in range(2,17):
    if 17 % i == 0:
        print(f'{17}不是質數,因為它可以被{i}除盡')
        break
    else:
        print(f'{17}是質數')
        break
```

```python
n = 17
isPrime = True
for i in range(2, n):
    if n % i == 0:
        print(f'{n}不是質數,因為它可以被{i}除盡')
        isPrime = False
        break

if isPrime==False:
    print(f'{n}是質數')
```
